```python
class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None
class DoublyLinkedList:
    def __init__(self):
        self.head = None
    def insert_first(self, data):
        new_node = Node(data)
        if self.head:
            self.head.prev = new_node
            new_node.next = self.head
        self.head = new_node
    def insert_last(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node
        new_node.prev = temp
    def insert_at_index(self, index, data):
        if index == 0:
            self.insert_first(data)
            return
        temp = self.head
        count = 0
        while temp and count < index - 1:
            temp = temp.next
            count += 1
        if temp is None:
            print("Index out of range")
            return
        new_node = Node(data)
        new_node.next = temp.next
        new_node.prev = temp
        if temp.next:
            temp.next.prev = new_node
        temp.next = new_node
    def delete_by_value(self, key):
        temp = self.head
        while temp:
            if temp.data == key:
                if temp.prev is None:
                    self.head = temp.next
                    if self.head:
                        self.head.prev = None
                else:
                    temp.prev.next = temp.next
                    if temp.next:
                        temp.next.prev = temp.prev
                print(key, "deleted")
                return
            temp = temp.next
        print("Element not found")
    def delete_by_index(self, index):
        if self.head is None:
            print("List is empty")
            return
        temp = self.head
        count = 0
        if index == 0:
            self.head = temp.next
            if self.head:
                self.head.prev = None
            print("Index", index, "deleted")
            return
        while temp and count < index:
            temp = temp.next
            count += 1
        if temp is None:
            print("Index out of range")
            return
        temp.prev.next = temp.next
        if temp.next:
            temp.next.prev = temp.prev
        print("Index", index, "deleted")
    def display(self):
        temp = self.head
        if temp is None:
            print("List is empty")
            return
        print("DLL:", end=" ")
        while temp:
            print(temp.data, end=" <-> ")
            temp = temp.next
        print("None")
dll = DoublyLinkedList()
 print("\n1.Insert First")
    print("2.Insert Last")
    print("3.Insert at Index")
    print("4.Delete by Element")
    print("5.Delete by Index")
    print("6.Display")
    print("7.Exit")
while True:
    choice = int(input("Enter your choice: "))
    if choice == 1:
        dll.insert_first(int(input("Enter value: ")))
    elif choice == 2:
        dll.insert_last(int(input("Enter value: ")))
    elif choice == 3:
        index = int(input("Enter index: "))
        data = int(input("Enter value: "))
        dll.insert_at_index(index, data)
    elif choice == 4:
        dll.delete_by_value(int(input("Enter element to delete: ")))
    elif choice == 5:
        dll.delete_by_index(int(input("Enter index to delete: ")))
    elif choice == 6:
        dll.display()
    elif choice == 7:
        break
    else:
        print("Invalid choice")
```
